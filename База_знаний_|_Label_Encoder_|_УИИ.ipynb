# Label Encoder

---



 При решении задач машинного обучения, часто приходится иметь дело со строковыми параметрами. Чаще всего это категориальные фичи, например названия городов, цвета, должность итд. Некоторые библиотеки могут изначально работать со строковыми параметрами, но для большинства предварительно надо перевести буквы в цифры. Более того, это даже очень рекомендуется т.к. в числовом виде будет использоваться меньше оперативной памяти для хранения данных для обучения. Процесс перевода строковых данных (labels) в числовые данные называется label encoding.

# Рассмотрим пример использования LabelEncoder для датасета headhunter

```python
# Библиотека для скачивания файлов из Google Drive
import gdown

url = "https://drive.google.com/uc?id=1kK9BaJ8rx84BTCEvfYZl9MIuknsiqmzT"  # Ссылка на датасет headhunter
output = "hh.csv"                                                            # Название файла после скачивания
gdown.download(url, output, quiet=True)                                      # Скачивание файла
```

```python
# Импортруем библиотеку Pandas
import pandas as pd

df = pd.read_csv("/content/hh.csv").head(5)  # Смотрим что находится в базе
schedule = df['График'].unique()             # Сохраняем текстовый формат колонки "График"
df.head(5)                                   # Выводится первые 5 строк датасета
```

В этом примере первый столбец (График) является полностью текстовым. Мы не можем использовать текст в данных для обучения модели. Поэтому, прежде чем мы сможем начать процесс, нам нужно эти данные подготовить.

 Для преобразования категорий в числовые данные используется класс LabelEncoder. Таким образом, чтобы получить признак для столбца "Занятость", импортируется класс из библиотеки sklearn и обработывается колонка функцией fit_transform.

```python
from sklearn import preprocessing                       # Импортируем модуль preprocessing в котором хранится LabelEncoder

label_encoder = preprocessing.LabelEncoder()            # Создаем объект label_encoder

df['График']= label_encoder.fit_transform(df['График']) # Кодируем колонку "График"

df.head(5)                                              # Выводится первые 5 строк датасета
```

Теперь, колонка "График" является закодированной с помощью LabelEncoder

```python
print(schedule)              # Текстовое представление признаков из колонки "График"
print(df['График'].unique()) # Закодированные значения для колонки "График"
```

Также, LabelEncoder можно использовать для одновременного кодирования нескольких столбцов.

```python
cols = ['Город', 'Занятость', 'Образование и ВУЗ', 'Ищет работу на должность'] # Названия колонок которые необходимо закодирвать

df[cols] = df[cols].apply(label_encoder.fit_transform)                        # Кодируются признаки сразу нескольких столбцов

df.head(5)                                                                     # Выводится первые 5 строк датасета
```
