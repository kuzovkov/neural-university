```python
# Ваше решение
# Ваше решение
import matplotlib.pyplot as plt    # Импорт модуля pyplot
import numpy as np                 # Импорт библиотеки numpy
import matplotlib

x = np.linspace(0, 20, 101)        # Генерация последовательности из 101 числа с шагом в 0.1
y = np.cos(x)                      # Запись cos(x) в переменную y
plt.figure(figsize=(12, 4))
plt.plot(x, y, label='cos(x)', c='#4b1234',marker='h', markerfacecolor='#ff00d5', markeredgewidth=2,
         markersize=10, markevery=5, linewidth=5)                     # Построение графика x, y

plt.xlabel('ось X', fontsize=8)
plt.ylabel('ось Y', fontsize=8)
plt.xticks(np.arange(0, 20.25, step=2.5), fontsize=8)
plt.yticks(np.arange(-1, 1.25, step=0.25), fontsize=8)
plt.title('cos(x)', fontsize=8)
plt.legend(loc='lower right')
plt.grid()
plt.show()                         # Фиксация графика
```
