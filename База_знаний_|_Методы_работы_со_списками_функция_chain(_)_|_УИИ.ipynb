Функция **chain( )** принадлежит модулю **itertools**, содержащему большое количество полезных инструментов для итерации.

```python
from itertools import chain
```

Функция **chain( )** принимает ряд объектов с итерируемыми элементами и возвращает один объект с объединённым количеством итерируемых элементов, взятых от каждого объекта. Данная функция способна объединить списки в один список, где она часто и применяется:

```python
# Возьмём два списка с разными элементами
numbers = [1, 2 , 3]
letters = ['a', 'b', 'c']
# Объединим списки через chain()
chained_list = list(chain(numbers, letters))
print(chained_list)
```

 Обратим внимание на тот факт, что тип данных **'cписок'** или **list()** задаётся совместно с применением функции **chain()**.

 Если в функцию **chain()** передать не списки, а строки - **string()**- то функция вернёт на выходе список с поэлементно разделёнными символами, из которых состояли данные строки:

```python
# В функцию chain() мы подаём две строки
greeting = list(chain('Hello', ' world!'))
print(greeting)
```

Если строковые значения являются элементами одного списка, то можно применить метод **chain.from_iterable()** для поэлементного разделения данных строк:

```python
string_letters = ['ABC', 'DEF', 'GHI']
sep_letters = list(chain.from_iterable(string_letters))
print(sep_letters)
```

В некоторых ситуациях схожие по функционалу методы работы со списками **extend( )** и **append( )** возвращают None type после присоединения к списку новых элементов. Это может создавать трудности при дальнейшей обработке данных. Функция **chain( )** удобна тем, что сохраняет тип объединяемых данных.
